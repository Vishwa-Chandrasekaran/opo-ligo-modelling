# Model 1 — Plane-Wave Parametric Amplification

## LIGO-style degenerate configuration

We begin with the simplest model of the nonlinear interaction:

$$
532~\mathrm{nm}
\rightarrow
1064~\mathrm{nm}+1064~\mathrm{nm}.
$$

The fields are labeled as:

- $A_1$: signal field at 1064 nm
- $A_2$: idler field at 1064 nm
- $A_3$: pump field at 532 nm

This notebook describes a **single-pass optical parametric amplifier (OPA)**.

It is not yet a complete optical parametric oscillator (OPO), because we do not include the cavity.

The purpose of this first model is to understand the relationship

$$
\boxed{
I_p
\rightarrow A_3
\rightarrow \kappa_1,\kappa_2
\rightarrow g
\rightarrow A_1(L),A_2(L)
\rightarrow G
}
$$

where $I_p$ is the pump intensity.

# 2. Coupled amplitude equations

For a second-order nonlinear interaction, the slowly varying amplitudes of the signal and idler satisfy coupled equations of the form

$$
\boxed{
\frac{dA_1}{dz}
=
\kappa_1 A_2^* e^{i\Delta k z}
}
$$

and

$$
\boxed{
\frac{dA_2}{dz}
=
\kappa_2 A_1^* e^{i\Delta k z}.
}
$$

The phase mismatch is

$$
\boxed{
\Delta k=k_3-k_1-k_2.
}
$$

The nonlinear coupling coefficients used in this model are

$$
\boxed{
\kappa_i=
\frac{2i\omega_i^2d_{\mathrm{eff}}A_3}
{k_i c^2}.
}
$$

Here:

- $d_{\mathrm{eff}}$ is the effective nonlinear coefficient.
- $A_3$ is the pump electric-field amplitude.
- $k_i$ is the wave number.
- $\omega_i$ is the angular frequency.
- $c$ is the speed of light.

# 3. Approximations used in Model 1

We deliberately make several approximations.

## 3.1 Slowly varying envelope approximation

The amplitude $A_i(z)$ varies slowly compared with the optical wavelength.

The rapidly varying phase is separated as

$$
E_i(z)=A_i(z)e^{ik_i z}.
$$

---

## 3.2 Undepleted pump approximation

The pump is assumed to remain approximately constant throughout the crystal:

$$
\boxed{
A_3(z)\approx A_3(0)=A_3.
}
$$

Therefore the nonlinear coupling coefficients $\kappa_1$ and $\kappa_2$ are constant along $z$.

---

## 3.3 Perfect phase matching

For the first model we assume

$$
\boxed{
\Delta k=0.
}
$$

Therefore,

$$
e^{i\Delta kz}=1.
$$

The coupled equations become

$$
\frac{dA_1}{dz}=\kappa_1A_2^*
$$

and

$$
\frac{dA_2}{dz}=\kappa_2A_1^*.
$$

---

## 3.4 Plane-wave approximation

The fields are treated as plane-wave-like modes.

There is no transverse spatial dependence.

Therefore:

$$
\boxed{
\text{No beam waist }w_0
}
$$

is used in this model.

Consequently, the natural pump input variable is **pump intensity**, rather than total pump power.

---

## 3.5 No cavity

There are no mirrors, cavity losses, output coupling, or round-trip propagation.

Therefore this is a **single-pass OPA model**, not yet an OPO model.

The OPO threshold will require the cavity to be included later.

# 4. Gain coefficient \(g\)

The coupled equations can be reduced to second-order differential equations.

The resulting gain parameter is

$$
\boxed{
g=
\left[
\kappa_1\kappa_2^*
-
\left(\frac{\Delta k}{2}\right)^2
\right]^{1/2}.
}
$$

For perfect phase matching,

$$
\Delta k=0,
$$

so

$$
\boxed{
g=\sqrt{\kappa_1\kappa_2^*}.
}
$$

Since

$$
\kappa_i\propto A_3,
$$

we have

$$
\kappa_1\kappa_2^*\propto |A_3|^2.
$$

Therefore,

$$
\boxed{
g\propto |A_3|.
}
$$

The quantity

$$
\boxed{gL}
$$

is dimensionless and represents the strength of the nonlinear interaction over a crystal of length $L$.

# 5. Solution for the signal and idler

For constant pump amplitude and perfect phase matching, the coupled equations have hyperbolic-function solutions.

If initially there is signal but no idler,

$$
\boxed{
A_2(0)=0,
}
$$

then the signal amplitude evolves as

$$
\boxed{
A_1(z)=A_1(0)\cosh(gz).
}
$$

The generated idler has the corresponding form

$$
A_2(z)\propto A_1^*(0)\sinh(gz).
$$

The exact phase factor multiplying the idler depends on the convention used for the complex coupling coefficients.

At the end of a crystal of length \(L\),

$$
\boxed{
A_1(L)=A_1(0)\cosh(gL).
}$$

# 6. Signal power gain

If the power is proportional to the squared field amplitude,

$$
P_1\propto |A_1|^2,
$$

then the signal power gain is

$$
G_s=
\frac{|A_1(L)|^2}{|A_1(0)|^2}.
$$

Using

$$
A_1(L)=A_1(0)\cosh(gL),
$$

we obtain

$$
\boxed{
G_s=\cosh^2(gL).
}
$$

Therefore the model predicts

$$
\boxed{
G_s(I_p)
=
\cosh^2\left[g(I_p)L\right].
}
$$

The next step is to connect the pump intensity $I_p$ to the pump field amplitude $A_3$.

# 7. Connecting pump intensity to pump field

For a plane electromagnetic wave propagating in a dielectric,

$$
\boxed{
I_p=
\frac{1}{2}n_p\epsilon_0c|A_3|^2.
}
$$

Therefore,

$$
\boxed{
|A_3|
=
\sqrt{
\frac{2I_p}
{n_p\epsilon_0c}
}.
}
$$

Thus,

$$
A_3\propto\sqrt{I_p}.
$$

Since

$$
g\propto A_3,
$$

we obtain

$$
\boxed{
g\propto\sqrt{I_p}.
}
$$

This predicts the qualitative shape of the first graph before we even run the numerical model.

# 8. Low-gain limit

When

$$
gL\ll1,
$$

we can use

$$
\cosh(gL)
\approx
1+\frac{(gL)^2}{2}.
$$

Therefore,

$$
G_s
=
\cosh^2(gL)
\approx
1+(gL)^2.
$$

Because

$$
g^2\propto I_p,
$$

we obtain

$$
\boxed{
G_s\approx1+C I_p
}
$$

for some constant $C$.

Therefore, at low pump intensity:

- $g$ follows approximately a square-root dependence on intensity.
- $G_s$ initially looks approximately linear in intensity.

At larger $gL$, the hyperbolic behavior becomes increasingly strong.

# 9. Parameters for the LIGO-style starting model

We use the degenerate configuration

$$
\boxed{
\lambda_p=532~\mathrm{nm}
}
$$

and

$$
\boxed{
\lambda_s=\lambda_i=1064~\mathrm{nm}.
}
$$

The nonlinear crystal is taken as PPKTP.

Starting values:

$$
d_{\mathrm{eff}}\approx10.7~\mathrm{pm/V}
$$

$$
L\approx10.55~\mathrm{mm}
$$

and representative refractive indices

$$
n_p\approx1.8896
$$

$$
n_s=n_i\approx1.8303.
$$

These are starting values for our model. Later we should replace the fixed refractive indices with wavelength- and temperature-dependent values.

The parameters $d_{\mathrm{eff}}$ and $L$ will be easy to vary in the code.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# ============================================================
# CONSTANTS
# ============================================================

c = 299792458.0                  # Speed of light [m/s]
epsilon0 = 8.8541878128e-12      # Vacuum permittivity [F/m]
```

```python
# ============================================================
# WAVELENGTHS
# ============================================================

lambda_p = 532e-9       # Pump wavelength [m]
lambda_s = 1064e-9      # Signal wavelength [m]
lambda_i = 1064e-9      # Idler wavelength [m]
```

```python
# ============================================================
# MATERIAL PARAMETERS
# ============================================================

# Representative PPKTP refractive indices

n_p = 1.8896            # 532 nm
n_s = 1.8303            # 1064 nm
n_i = 1.8303            # 1064 nm

# Effective nonlinear coefficient

d_eff = 10.7e-12        # [m/V]

# Crystal length

L = 10.55e-3            # [m]
```

```python
# ============================================================
# PHASE MATCHING
# ============================================================

# Model 1 assumes perfect phase matching

delta_k = 0.0           # [1/m]
```

```python
# ============================================================
# PUMP INTENSITY
# ============================================================

# Pump intensity range [W/m^2]

I_pump = np.linspace(1e3, 1e10, 1000)
```

```python
# ============================================================
# ANGULAR FREQUENCIES
# ============================================================

omega_p = 2 * np.pi * c / lambda_p
omega_s = 2 * np.pi * c / lambda_s
omega_i = 2 * np.pi * c / lambda_i
```

```python
# ============================================================
# WAVE NUMBERS
# ============================================================

k_p = n_p * omega_p / c
k_s = n_s * omega_s / c
k_i = n_i * omega_i / c
```

```python
# ============================================================
# PUMP FIELD AMPLITUDE
# ============================================================

# For a plane wave:
#
# I = (1/2) n epsilon0 c |A3|^2
#
# Therefore:
#
# |A3| = sqrt(2 I / (n epsilon0 c))

A3 = np.sqrt(
    2 * I_pump /
    (n_p * epsilon0 * c)
)
```

```python
# ============================================================
# NONLINEAR COUPLING COEFFICIENTS
# ============================================================

# Your definition:
#
# kappa_i =
# 2 i omega_i^2 d_eff A3
# ----------------------
#        k_i c^2

kappa_1 = (
    2j * omega_s**2 * d_eff * A3
    / (k_s * c**2)
)

kappa_2 = (
    2j * omega_i**2 * d_eff * A3
    / (k_i * c**2)
)
```

```python
# ============================================================
# PARAMETRIC GAIN COEFFICIENT g
# ============================================================

# g = sqrt[
#       kappa_1 kappa_2*
#       - (delta_k / 2)^2
#     ]

g = np.sqrt(
    kappa_1 * np.conjugate(kappa_2)
    - (delta_k / 2)**2
)
```

```python
# ============================================================
# SIGNAL GAIN
# ============================================================

# A1(L) = A1(0) cosh(gL)
#
# Therefore:
#
# G = |A1(L) / A1(0)|^2
#   = cosh^2(gL)

G_signal = np.abs(
    np.cosh(g * L)
)**2
```

# 10. Inspect the numerical values

Before looking at the plots, it is useful to inspect the magnitude of

$$
A_3,\qquad g,\qquad gL
$$

and the resulting gain.

The quantity $gL$ is especially important because it determines how strongly the fields are amplified.

```python
print("Pump field amplitude")
print("--------------------")
print(f"Minimum |A3| = {np.min(np.abs(A3)):.3e} V/m")
print(f"Maximum |A3| = {np.max(np.abs(A3)):.3e} V/m")

print("\nParametric gain coefficient")
print("---------------------------")
print(f"Minimum g = {np.min(np.real(g)):.3e} 1/m")
print(f"Maximum g = {np.max(np.real(g)):.3e} 1/m")

print("\nDimensionless interaction strength")
print("-----------------------------------")
print(f"Minimum gL = {np.min(np.real(g * L)):.3e}")
print(f"Maximum gL = {np.max(np.real(g * L)):.3e}")

print("\nSignal power gain")
print("-----------------")
print(f"Minimum G = {np.min(G_signal):.6f}")
print(f"Maximum G = {np.max(G_signal):.6f}")
```

# 11. Plot \(g\) versus pump intensity

From the analytical model,

$$
g\propto\sqrt{I_p}.
$$

Therefore we expect a square-root-like curve.

```python
plt.figure(figsize=(8, 5))

plt.plot(I_pump, np.real(g))

plt.xlabel("Pump intensity [W/m²]")
plt.ylabel("g [1/m]")
plt.title("Parametric Gain Coefficient vs Pump Intensity")

plt.grid(True)
plt.tight_layout()

plt.show()
```

# 12. Plot signal power gain

The signal gain is

$$
\boxed{
G_s=\cosh^2(gL).
}
$$

At low pump intensity, where

$$
gL\ll1,
$$

the gain remains close to unity.

As $gL$ becomes larger, the nonlinear amplification becomes increasingly pronounced.

```python
plt.figure(figsize=(8, 5))

plt.plot(I_pump, G_signal)

plt.xlabel("Pump intensity [W/m²]")
plt.ylabel("Signal power gain")
plt.title("Signal Gain vs Pump Intensity")

plt.grid(True)
plt.tight_layout()

plt.show()
```

# 13. Plot the dimensionless interaction strength \(gL\)

The quantity

$$
gL
$$

is useful for understanding the regime of the interaction.

- $gL\ll1$: weak interaction
- $gL\sim0.1$: small but noticeable amplification
- $gL\sim1$: strong single-pass parametric amplification

These are qualitative guideposts rather than strict physical boundaries.

```python
plt.figure(figsize=(8, 5))

plt.plot(I_pump, np.real(g * L))

plt.xlabel("Pump intensity [W/m²]")
plt.ylabel("gL")
plt.title("Nonlinear Interaction Strength vs Pump Intensity")

plt.grid(True)
plt.tight_layout()

plt.show()
```

# 14. Varying the nonlinear coefficient

From

$$
\kappa_i\propto d_{\mathrm{eff}},
$$

we have

$$
g\propto d_{\mathrm{eff}}.
$$

Therefore increasing $d_{\mathrm{eff}}$ increases the parametric gain.

```python
# Example: compare several nonlinear coefficients

d_values = [5e-12, 10e-12, 15e-12, 20e-12]

plt.figure(figsize=(8, 5))

for d in d_values:

    kappa_1_test = (
        2j * omega_s**2 * d * A3
        / (k_s * c**2)
    )

    kappa_2_test = (
        2j * omega_i**2 * d * A3
        / (k_i * c**2)
    )

    g_test = np.sqrt(
        kappa_1_test *
        np.conjugate(kappa_2_test)
    )

    G_test = np.abs(
        np.cosh(g_test * L)
    )**2

    plt.plot(
        I_pump,
        G_test,
        label=f"d_eff = {d*1e12:.1f} pm/V"
    )

plt.xlabel("Pump intensity [W/m²]")
plt.ylabel("Signal power gain")
plt.title("Effect of Nonlinear Coefficient")

plt.legend()
plt.grid(True)
plt.tight_layout()

plt.show()
```

# 15. Varying crystal length

The gain depends on

$$
gL.
$$

Therefore, for a fixed pump intensity,

$$
\boxed{
\text{longer crystal}
\rightarrow
\text{larger interaction}
\rightarrow
\text{larger gain}.
}
$$

This is the simple plane-wave picture.

In a realistic **Gaussian-beam model**, however, increasing the crystal length is not always equivalent to increasing the useful nonlinear interaction indefinitely. **Focusing, beam divergence, phase matching, absorption and spatial mode effects eventually become important.**

```python
# Example: compare several crystal lengths

L_values = [5e-3, 10e-3, 15e-3, 20e-3]

plt.figure(figsize=(8, 5))

for L_test in L_values:

    G_test = np.abs(
        np.cosh(g * L_test)
    )**2

    plt.plot(
        I_pump,
        G_test,
        label=f"L = {L_test*1e3:.1f} mm"
    )

plt.xlabel("Pump intensity [W/m²]")
plt.ylabel("Signal power gain")
plt.title("Effect of Crystal Length")

plt.legend()
plt.grid(True)
plt.tight_layout()

plt.show()
```

# 16. Summary of Model 1

The complete chain is

$$
\boxed{
I_p
\rightarrow
A_3
\rightarrow
\kappa_1,\kappa_2
\rightarrow
g
\rightarrow
A_1(L)
\rightarrow
G_s
}
$$

with

$$
\boxed{
A_3=
\sqrt{
\frac{2I_p}
{n_p\epsilon_0c}
}
}
$$

$$
\boxed{
\kappa_i=
\frac{2i\omega_i^2d_{\mathrm{eff}}A_3}
{k_i c^2}
}
$$

$$
\boxed{
g=
\sqrt{
\kappa_1\kappa_2^*
-
\left(\frac{\Delta k}{2}\right)^2
}
}
$$

and, for perfect phase matching,

$$
\boxed{
g=\sqrt{\kappa_1\kappa_2^*}
}
$$

followed by

$$
\boxed{
A_1(L)=A_1(0)\cosh(gL)
}
$$

and

$$
\boxed{
G_s=\cosh^2(gL).
}
$$

# 17. What we have NOT modeled yet

This model intentionally does not include:

- Gaussian transverse profile
- beam waist \(w_0\)
- focusing
- Gouy phase
- pump depletion
- phase mismatch
- cavity mirrors
- cavity losses
- output coupling
- OPO threshold
- vacuum input
- squeezing

These should be introduced progressively rather than all at once.

## Next model

The next model will replace the plane-wave pump with a Gaussian beam.

Then we will use

$$
I(r)=I_0e^{-2r^2/w_0^2}
$$

and

$$
P_p=\frac{\pi w_0^2}{2}I_0
$$

to connect the **actual pump power** to the peak pump field.

That will give us

$$
\boxed{
P_p,w_0
\rightarrow
A_3
\rightarrow
\kappa_i
\rightarrow
g.
}
$$

Only after understanding that model should we move to the cavity and calculate the actual **OPO threshold**.
